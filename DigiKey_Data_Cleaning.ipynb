```python
ps_clean = ps.copy()
```

```python
keep_columns = [
    "Digi-Key Part Number",
    "Manufacturer Part Number",
    "Manufacturer",
    "Description",
    "Quantity Available",
    "Unit Price (USD)",
    "Product Category",
    "Package / Case",
    "RoHS Status",
    "Minimum Order Qty",
    "Currency",
    "Product URL",
    "Datasheet URL",
    "Last Updated",
    "Warehouse"
]
```

```python
ps_clean = ps_clean[keep_columns].copy()
```

```python
duplicates = ps_clean[ps_clean.duplicated(keep=False)]

duplicates.head(10)
```

```python
rows_before = len(ps_clean)

ps_clean = ps_clean.drop_duplicates()

rows_after = len(ps_clean)

print("Rows before cleaning:", rows_before)
print("Rows after cleaning:", rows_after)
print("Duplicate rows removed:", rows_before - rows_after)
```

```python
ps_clean.isna().sum()
```

```python
missing_summary = pd.DataFrame({
    "Missing Values": ps_clean.isna().sum(),
    "Missing Percentage": (ps_clean.isna().mean() * 100).round(2)
})

missing_summary.sort_values(
    by="Missing Percentage",
    ascending=False
)
```

```python
ps_clean = ps_clean.replace(r'^\s*$', pd.NA, regex=True)
```

```python
ps_clean.isna().sum().sort_values(ascending=False)
```

```python
ps_clean["Digi-Key Part Number"].isna().sum()
```

```python
ps_clean.loc[
    ps_clean["Digi-Key Part Number"].isna(),
    [
        "Digi-Key Part Number",
        "Manufacturer Part Number",
        "Manufacturer",
        "Description"
    ]
].head(10)
```

```python
ps_clean["Digi-Key Part Number"].isna().sum()
```

```python
ps_clean["Missing Digi-Key Part Number"] = (
    ps_clean["Digi-Key Part Number"].isna()
)
```

```python
ps_clean[
    ps_clean["Missing Digi-Key Part Number"]
].head(10)
```

```python
# Remove unwanted spaces, dollar signs and commas
ps_clean["Unit Price (USD)"] = (
    ps_clean["Unit Price (USD)"]
    .astype("string")
    .str.strip()
    .str.replace(r"[$,]", "", regex=True)
)

# Convert price values into numeric format
ps_clean["Unit Price (USD)"] = pd.to_numeric(
    ps_clean["Unit Price (USD)"],
    errors="coerce"
)
```

```python
ps_clean["Quantity Available"] = (
    ps_clean["Quantity Available"]
    .astype("string")
    .str.strip()
    .str.replace(",", "", regex=False)
)

ps_clean["Quantity Available"] = pd.to_numeric(
    ps_clean["Quantity Available"],
    errors="coerce"
)
```

```python
ps_clean[["Unit Price (USD)", "Quantity Available"]].dtypes
```

```python
ps_clean[["Unit Price (USD)", "Quantity Available"]].isna().sum()
```

```python
missing_price_indexes = ps_clean.index[
    ps_clean["Unit Price (USD)"].isna()
]

ps.loc[
    missing_price_indexes,
    [
        "Digi-Key Part Number",
        "Manufacturer Part Number",
        "Unit Price (USD)"
    ]
].head(15)
```

```python
# Identify text columns
text_columns = ps_clean.select_dtypes(
    include=["object", "string"]
).columns

# Remove extra spaces and standardize missing values
for col in text_columns:
    ps_clean[col] = ps_clean[col].astype("string").str.strip()

    ps_clean[col] = ps_clean[col].replace(
        [
            "", "N/A", "NA", "n/a", "na",
            "NaN", "nan", "None", "null",
            "<NA>", "TBD", "tbd"
        ],
        pd.NA
    )

print("Text cleaning completed!")
```

```python
duplicate_parts_count = (
    ps_clean["Digi-Key Part Number"]
    .dropna()
    .duplicated()
    .sum()
)

print("Duplicate Part Numbers:", duplicate_parts_count)
```

```python
duplicate_parts = ps_clean[
    ps_clean["Digi-Key Part Number"].notna()
    & ps_clean["Digi-Key Part Number"].duplicated(keep=False)
].sort_values("Digi-Key Part Number")

duplicate_parts.head(20)
```

```python
ps_clean.columns = (
    ps_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

ps_clean.columns.tolist()
```
